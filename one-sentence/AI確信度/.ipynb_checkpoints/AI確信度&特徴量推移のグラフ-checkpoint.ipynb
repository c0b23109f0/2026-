```python

import os
import re
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

# ==========================================
# 1. CSV読み込み
# ==========================================
csv_path = os.path.join(
    current_dir,
    "LR_predict_proba_cumulative_features_merged.csv"
)

df = pd.read_csv(csv_path)

print("データサイズ:", df.shape)

# ==========================================
# 2. 特徴量を5グループに分類
# ==========================================
feature_groups = {
    "機能語": [
        "機能語率", "助詞割合", "助動詞割合",
        "連体詞割合", "接続詞割合",
        "助詞_と", "助詞_は", "助詞_の", "助詞_で",
        "助詞_て", "助詞_に", "助詞_や", "助詞_を",
        "助詞_ば", "助詞_が", "助詞_も", "助詞_つつ",
        "助詞_から", "助詞_など", "助詞_へ", "助詞_まで",
        "助詞_か", "助詞_たり", "助詞_だけ",
        "助詞_ながら", "助詞_より"
    ],

    "内容語": [
        "内容語率", "名詞割合", "動詞割合",
        "形容詞割合", "形状詞割合", "副詞割合",
        "感動詞割合"
    ],

    "語彙多様性・反復": [
        "異なり語数", "TTR",
        "Hapax率_all", "Yules_K_all",
        "Hapax率_func", "Yules_K_func",
        "Hapax率_cont", "Yules_K_cont"
    ],

    "文長・文構造": [
        "総単語数", "文長標準偏差",
        "最大文長", "平均文長"
    ],

    "表記・句読点": [
        "ひらがな率", "カタカナ率",
        "漢字率", "英数字率",
        "読点頻度", "句点頻度",
        "感嘆符数", "疑問符数"
    ]
}

# ==========================================
# 3. 保存先フォルダ
# ==========================================
output_dir = os.path.join(
    current_dir,
    "LR_特徴量グループ別グラフ"
)
os.makedirs(output_dir, exist_ok=True)

# 特徴量以外の列
meta_cols = {
    "doc_id", "label", "sent_order", "AI確率"
}

# 各グループの存在確認
for group_name, features in feature_groups.items():
    available = [f for f in features if f in df.columns]
    missing = [f for f in features if f not in df.columns]

    print(f"\n【{group_name}】")
    print(f"対象特徴量: {len(available)} / {len(features)}")

    if missing:
        print("CSVに存在しない列:", missing)

# ==========================================
# 4. 文章ごと・グループごとに描画
# ==========================================
doc_ids = df["doc_id"].dropna().unique()

for doc_id in doc_ids:
    data = df[df["doc_id"] == doc_id].copy()
    data = data.sort_values("sent_order")

    if data.empty:
        continue

    label = data["label"].iloc[0]

    for group_name, features in feature_groups.items():

        # CSVに実在する特徴量だけ使用
        available = [f for f in features if f in data.columns]

        if not available:
            continue

        # 数値に変換
        plot_data = data[available].apply(
            pd.to_numeric, errors="coerce"
        )

        # 変動がない特徴量や全欠損の特徴量は除外
        valid_features = []
        z_data = pd.DataFrame(index=data.index)

        for feature in available:
            values = plot_data[feature]
            std = values.std(ddof=0)

            if (
                values.notna().sum() < 2
                or pd.isna(std)
                or std == 0
            ):
                continue

            z_data[feature] = (
                values - values.mean()
            ) / std

            valid_features.append(feature)

        # ==================================
        # グラフ
        # ==================================
        fig, (ax1, ax2) = plt.subplots(
            2, 1,
            figsize=(16, max(9, len(valid_features) * 0.35)),
            sharex=True,
            gridspec_kw={"height_ratios": [1, 2]}
        )

        # 上段：AI確率
        ax1.plot(
            data["sent_order"],
            data["AI確率"],
            marker="o",
            linewidth=2.5,
            color="#ff7f0e",
            label="AI確率"
        )

        ax1.axhline(
            0.5, color="black",
            linestyle=":", alpha=0.7,
            label="判定境界（0.5）"
        )

        ax1.set_ylabel("AI確率", fontsize=12)
        ax1.set_ylim(-0.05, 1.05)
        ax1.set_title(
            f"{label}：{doc_id}｜{group_name}",
            fontsize=15
        )
        ax1.grid(True, linestyle=":", alpha=0.5)
        ax1.legend(loc="best")

        # 下段：グループ内の特徴量
        for feature in valid_features:
            ax2.plot(
                data["sent_order"],
                z_data[feature],
                marker="o",
                linewidth=1.8,
                markersize=3,
                label=feature
            )

        ax2.axhline(
            0, color="black",
            linestyle=":", alpha=0.6
        )

        ax2.set_xlabel("入力文章数", fontsize=12)
        ax2.set_ylabel("特徴量のZスコア", fontsize=12)
        ax2.set_xticks(sorted(data["sent_order"].unique()))
        ax2.grid(True, linestyle=":", alpha=0.4)

        # 特徴量が多い場合も凡例を読みやすくする
        if valid_features:
            ax2.legend(
                loc="upper left",
                bbox_to_anchor=(1.01, 1),
                fontsize=9
            )

        plt.tight_layout()

        # ファイル名に使えない文字を置換
        safe_doc_id = re.sub(
            r'[\\/:*?"<>|]', "_", str(doc_id)
        )
        safe_group = re.sub(
            r'[\\/:*?"<>|]', "_", group_name
        )

        output_png = os.path.join(
            output_dir,
            f"{safe_doc_id}_{safe_group}.png"
        )

        plt.savefig(
            output_png,
            dpi=150,
            bbox_inches="tight"
        )
        plt.close(fig)

print("\nすべてのグラフを保存しました。")
print("保存先:", output_dir)
```
