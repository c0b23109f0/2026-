```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 1. サンプルデータの用意 (実際のデータに合わせて置き換えてください)
# ※ 日本語の場合は事前に MeCab などで形態素解析（分かち書き）しておく必要があります。
corpus = [
    "これは とても 素晴らしい 商品 です",
    "本当に 最悪な 経験 を しました",
    "普通 の クオリティ で 特に見る べき ところ は ない",
    "大満足 です また 買い ます",
    "お金 の 無駄 でした 二度と 買いません",
    "可もなく不可もなく という 感じ です"
]
# 正解ラベル (1: ポジティブ, 0: ネガティブ)
labels = [1, 0, 0, 1, 0, 0]

# 2. データの分割 (訓練データとテストデータ)
X_train, X_test, y_train, y_test = train_test_split(corpus, labels, test_size=0.3, random_state=42)

# 3. TF-IDF によるテキストのベクトル化
# ベクトルの最大特徴量数(max_features)などはタスクに合わせて調整してください
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# 4. ロジスティック回帰モデルの学習
clf = LogisticRegression(random_state=42)
clf.fit(X_train_tfidf, y_train)

# 5. 予測と評価
y_pred = clf.predict(X_test_tfidf)

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
