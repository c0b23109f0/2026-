```python
import tiktoken

text = "今日の天気は快晴です。"

enc = tiktoken.get_encoding("cl100k_base")

tokens = enc.encode(text)

print(tokens)

print("\n=== BPE Tokens ===")

for token in tokens:
    piece = enc.decode([token])
    print(repr(piece))
```

```python
from collections import Counter
import tiktoken

with open("Human_text_01.txt", "r", encoding="utf-8") as file:
    text = file.read()

sentences = text.split("。")
sentences = [s for s in sentences if s.strip()]

bpe_sentence_lengths = []
    
enc = tiktoken.get_encoding("cl100k_base")

tokens = enc.encode(text)

bpe_count = len(tokens)

unique_bpe_count = len(set(tokens))

bpe_ttr = unique_bpe_count / bpe_count

decode_tokens = []

for token in tokens:
    try:
        piece = enc.decode([token])
        decode_tokens.append(piece)
    except:
        pass
    
for sentence in sentences:
    tokens = enc.encode(sentence)
    bpe_sentence_lengths.append(len(tokens))
    
avg_bpe_sentence_length = (
    sum(bpe_sentence_lengths)
    / len(bpe_sentence_lengths)
)

token_freq = Counter(tokens)

top10 = token_freq.most_common(10)

print("=== BPE 特徴量 ===")

print("総BPE数：", bpe_count)

print("異なりBPE数：", unique_bpe_count)

print("BPE TTR：", round(bpe_ttr, 4))

print("平均BPE文長：", round(avg_bpe_sentence_length, 4))

print("\n=== 頻出BPE Top10 ===")

for token_id, freq in top10:
    try:
        token_text = enc.decode([token_id])
    except:
        token_text = "[Decode Error]"
        
    print(
        f"ID:{token_id}-"
        f"Token:{repr(token_text)}-"
        f"Freq:{freq}"
    )
print(len(sentences))
```
